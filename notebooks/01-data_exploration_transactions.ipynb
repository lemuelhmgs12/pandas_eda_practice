### Observations

- **Shape:** 38,506 rows × 6 columns
- **Date range:** April 1 – September 30, 2020 (6 months of transactions)
- **Customers:** IDs range from 1–870, suggesting ~870 unique customers
- **Products:** 5 product areas (`product_area_id` 1–5 )
- **Items per transaction:** Median 4 items, max 310 — the max looks like an outlier worth investigating
- **Sales cost:** Median $23.18, mean $40.25 — mean is pulled up by high-value outliers (max $669.34). Min is $0.00 — worth checking if those are returns or data errors
- **Missing values:** None detected across all 6 columns (count = 38,506 for all)
