## 4. Sorting & Outlier Investigation

Sorting by `distance_from_store` ascending to surface nulls and outliers.

```python
# sorting

sorted_customers = customer_details.sort_values(
    by='distance_from_store', 
    inplace=False, 
    ascending= True, 
    na_position='first')
```

```python
sorted_customers.head(10)
```

```python
sorted_customers.tail()
```

### Observations

- **Shape:** 870 rows × 4 columns (870 unique customers)
- **Missing values:** `distance_from_store` 5 nulls, `gender` 5 nulls, `credit_score` 8 nulls — small enough to impute or drop depending on use case
- **Distance from store:** Median 1.66 miles but max is 400.97 — clear outlier, mean (2.61) is heavily skewed by it
- **Credit score:** Ranges 0.26–0.88, mean 0.60 — appears to be a normalized score (0–1 scale)
- **Gender:** 2 unique values (M/F), no unexpected categories
- **Distance from store:** Median 1.66 miles but max 400.97 is a significant outlier — next highest is 113.89, likely a data entry error. Mean (2.61) is skewed by it
- **3 customers at 0.00 miles** — possible store employees or a data issue, worth flagging
- **Nulls surface at top when sorted** — consistent with `isna().sum()` output (5 nulls in `distance_from_store`)
